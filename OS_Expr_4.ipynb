```python
%%writefile cpu_scheduling.c
#include <stdio.h>

int main() {
    int n, i;
    int burstTime[20], waitingTime[20], turnaroundTime[20];
    float avgWaiting = 0, avgTurnaround = 0;

    printf("Enter number of processes: ");
    scanf("%d", &n);

    printf("Enter burst time for each process:\n");

    for (i = 0; i < n; i++) {
        printf("P%d: ", i + 1);
        scanf("%d", &burstTime[i]);
    }

    // First process has zero waiting time
    waitingTime[0] = 0;

    // Calculate waiting time
    for (i = 1; i < n; i++) {
        waitingTime[i] = waitingTime[i - 1] + burstTime[i - 1];
    }

    // Calculate turnaround time
    for (i = 0; i < n; i++) {
        turnaroundTime[i] = waitingTime[i] + burstTime[i];

        avgWaiting += waitingTime[i];
        avgTurnaround += turnaroundTime[i];
    }

    printf("\nProcess\tBurst Time\tWaiting Time\tTurnaround Time\n");

    for (i = 0; i < n; i++) {
        printf("P%d\t%d\t\t%d\t\t%d\n",
               i + 1,
               burstTime[i],
               waitingTime[i],
               turnaroundTime[i]);
    }

    avgWaiting /= n;
    avgTurnaround /= n;

    printf("\nAverage Waiting Time = %.2f\n", avgWaiting);
    printf("Average Turnaround Time = %.2f\n", avgTurnaround);

    return 0;
}
```

```python
%%writefile cpu_scheduling.sh
#!/bin/bash

echo "CPU Scheduling - FCFS Algorithm"
echo "--------------------------------"

read -p "Enter number of processes: " n

declare -a burst
declare -a waiting
declare -a turnaround

echo "Enter burst time for each process:"

for ((i=0; i<n; i++))
do
    read -p "P$((i+1)): " burst[$i]
done

# First process has zero waiting time
waiting[0]=0

# Calculate waiting time
for ((i=1; i<n; i++))
do
    waiting[$i]=$((waiting[$i-1] + burst[$i-1]))
done

# Calculate turnaround time
totalWaiting=0
totalTurnaround=0

for ((i=0; i<n; i++))
do
    turnaround[$i]=$((waiting[$i] + burst[$i]))

    totalWaiting=$((totalWaiting + waiting[$i]))
    totalTurnaround=$((totalTurnaround + turnaround[$i]))
done

echo ""
echo -e "Process\tBurst Time\tWaiting Time\tTurnaround Time"

for ((i=0; i<n; i++))
do
    echo -e "P$((i+1))\t${burst[$i]}\t\t${waiting[$i]}\t\t${turnaround[$i]}"
done

echo ""
echo "Average Waiting Time = $(awk "BEGIN {printf \"%.2f\", $totalWaiting/$n}")"
echo "Average Turnaround Time = $(awk "BEGIN {printf \"%.2f\", $totalTurnaround/$n}")"
```
