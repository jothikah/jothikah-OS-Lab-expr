```python
%%writefile ipc.c
#include <stdio.h>
#include <stdlib.h>
#include <unistd.h>
#include <sys/wait.h>
#include <string.h>

int main() {
    int fd[2];
    pid_t pid;
    char message[] = "Hello from Parent Process!";
    char buffer[100];

    // Create pipe
    if (pipe(fd) == -1) {
        perror("Pipe creation failed");
        exit(1);
    }

    // Create child process
    pid = fork();

    if (pid < 0) {
        perror("Fork failed");
        exit(1);
    }

    if (pid > 0) {
        // Parent process
        close(fd[0]);  // Close reading end

        printf("Parent Process (PID: %d)\n", getpid());
        printf("Sending message to child...\n");

        write(fd[1], message, strlen(message) + 1);

        close(fd[1]);  // Close writing end

        wait(NULL);

        printf("Parent process completed.\n");
    }
    else {
        // Child process
        close(fd[1]);  // Close writing end

        read(fd[0], buffer, sizeof(buffer));

        printf("Child Process (PID: %d)\n", getpid());
        printf("Message received: %s\n", buffer);

        close(fd[0]);  // Close reading end

        exit(0);
    }

    return 0;
}
```

```python
!gcc ipc.c -o ipc
!./ipc
```

```python
%%writefile ipc.sh
#!/bin/bash

echo "Inter Process Communication using Shell"
echo "----------------------------------------"

echo "Message from Process 1"
echo "Hello from Process 1" | grep "Hello"

echo ""
echo "Process communication using pipe:"
echo "UNIX IPC Example" | tr 'a-z' 'A-Z'

echo ""
echo "Sending data from one process to another:"
echo "Hello Child Process" | while read message
do
    echo "Child received: $message"
done

echo ""
echo "IPC demonstration completed."
```
