```python
%%writefile paging.c
#include <stdio.h>

int main() {
    int pageSize, numPages;
    int pageTable[20];
    int logicalAddress;
    int pageNumber, offset, frameNumber, physicalAddress;

    printf("Paging Technique\n");
    printf("----------------\n");

    printf("Enter page size: ");
    scanf("%d", &pageSize);

    printf("Enter number of pages: ");
    scanf("%d", &numPages);

    printf("\nEnter the page table:\n");

    for (int i = 0; i < numPages; i++) {
        printf("Page %d -> Frame: ", i);
        scanf("%d", &pageTable[i]);
    }

    printf("\nEnter logical address: ");
    scanf("%d", &logicalAddress);

    // Calculate page number and offset
    pageNumber = logicalAddress / pageSize;
    offset = logicalAddress % pageSize;

    if (pageNumber >= numPages) {
        printf("Invalid logical address!\n");
        return 1;
    }

    // Get frame number from page table
    frameNumber = pageTable[pageNumber];

    // Calculate physical address
    physicalAddress = (frameNumber * pageSize) + offset;

    printf("\nLogical Address = %d\n", logicalAddress);
    printf("Page Number = %d\n", pageNumber);
    printf("Offset = %d\n", offset);
    printf("Frame Number = %d\n", frameNumber);
    printf("Physical Address = %d\n", physicalAddress);

    return 0;
}
```

```python
%%writefile paging.sh
#!/bin/bash

echo "Paging Technique"
echo "----------------"

read -p "Enter page size: " pageSize
read -p "Enter number of pages: " numPages

declare -a pageTable

echo ""
echo "Enter the page table:"

for ((i=0; i<numPages; i++))
do
    read -p "Page $i -> Frame: " pageTable[$i]
done

echo ""
read -p "Enter logical address: " logicalAddress

# Calculate page number and offset
pageNumber=$((logicalAddress / pageSize))
offset=$((logicalAddress % pageSize))

if [ $pageNumber -ge $numPages ]
then
    echo "Invalid logical address!"
    exit 1
fi

# Get frame number
frameNumber=${pageTable[$pageNumber]}

# Calculate physical address
physicalAddress=$((frameNumber * pageSize + offset))

echo ""
echo "Logical Address = $logicalAddress"
echo "Page Number = $pageNumber"
echo "Offset = $offset"
echo "Frame Number = $frameNumber"
echo "Physical Address = $physicalAddress"
```

```python
!chmod +x paging.sh
!./paging.sh
```
