```python
%%writefile memory.c
#include <stdio.h>

void firstFit(int blocks[], int b, int processes[], int p) {
    int allocation[p];

    for (int i = 0; i < p; i++)
        allocation[i] = -1;

    for (int i = 0; i < p; i++) {
        for (int j = 0; j < b; j++) {
            if (blocks[j] >= processes[i]) {
                allocation[i] = j;
                blocks[j] -= processes[i];
                break;
            }
        }
    }

    printf("\n--- First Fit ---\n");
    printf("Process\tSize\tBlock\n");

    for (int i = 0; i < p; i++) {
        if (allocation[i] != -1)
            printf("P%d\t%d\tB%d\n",
                   i + 1, processes[i], allocation[i] + 1);
        else
            printf("P%d\t%d\tNot Allocated\n",
                   i + 1, processes[i]);
    }
}

void bestFit(int blocks[], int b, int processes[], int p) {
    int allocation[p];

    for (int i = 0; i < p; i++)
        allocation[i] = -1;

    for (int i = 0; i < p; i++) {
        int best = -1;

        for (int j = 0; j < b; j++) {
            if (blocks[j] >= processes[i]) {
                if (best == -1 || blocks[j] < blocks[best])
                    best = j;
            }
        }

        if (best != -1) {
            allocation[i] = best;
            blocks[best] -= processes[i];
        }
    }

    printf("\n--- Best Fit ---\n");
    printf("Process\tSize\tBlock\n");

    for (int i = 0; i < p; i++) {
        if (allocation[i] != -1)
            printf("P%d\t%d\tB%d\n",
                   i + 1, processes[i], allocation[i] + 1);
        else
            printf("P%d\t%d\tNot Allocated\n",
                   i + 1, processes[i]);
    }
}

void worstFit(int blocks[], int b, int processes[], int p) {
    int allocation[p];

    for (int i = 0; i < p; i++)
        allocation[i] = -1;

    for (int i = 0; i < p; i++) {
        int worst = -1;

        for (int j = 0; j < b; j++) {
            if (blocks[j] >= processes[i]) {
                if (worst == -1 || blocks[j] > blocks[worst])
                    worst = j;
            }
        }

        if (worst != -1) {
            allocation[i] = worst;
            blocks[worst] -= processes[i];
        }
    }

    printf("\n--- Worst Fit ---\n");
    printf("Process\tSize\tBlock\n");

    for (int i = 0; i < p; i++) {
        if (allocation[i] != -1)
            printf("P%d\t%d\tB%d\n",
                   i + 1, processes[i], allocation[i] + 1);
        else
            printf("P%d\t%d\tNot Allocated\n",
                   i + 1, processes[i]);
    }
}

int main() {
    int b, p;

    printf("Memory Allocation Methods\n");
    printf("-------------------------\n");

    printf("Enter number of memory blocks: ");
    scanf("%d", &b);

    int blocks[b], original[b];

    printf("Enter size of each memory block:\n");
    for (int i = 0; i < b; i++) {
        printf("B%d: ", i + 1);
        scanf("%d", &blocks[i]);
        original[i] = blocks[i];
    }

    printf("\nEnter number of processes: ");
    scanf("%d", &p);

    int processes[p];

    printf("Enter size of each process:\n");
    for (int i = 0; i < p; i++) {
        printf("P%d: ", i + 1);
        scanf("%d", &processes[i]);
    }

    // First Fit
    for (int i = 0; i < b; i++)
        blocks[i] = original[i];

    firstFit(blocks, b, processes, p);

    // Best Fit
    for (int i = 0; i < b; i++)
        blocks[i] = original[i];

    bestFit(blocks, b, processes, p);

    // Worst Fit
    for (int i = 0; i < b; i++)
        blocks[i] = original[i];

    worstFit(blocks, b, processes, p);

    return 0;
}
```

```python
%%writefile memory.sh
#!/bin/bash

echo "Memory Allocation Methods"
echo "-------------------------"

read -p "Enter number of memory blocks: " b

declare -a blocks
declare -a original

echo "Enter size of each memory block:"

for ((i=0; i<b; i++))
do
    read -p "B$((i+1)): " blocks[$i]
    original[$i]=${blocks[$i]}
done

read -p "Enter number of processes: " p

declare -a process

echo "Enter size of each process:"

for ((i=0; i<p; i++))
do
    read -p "P$((i+1)): " process[$i]
done

# ---------------- FIRST FIT ----------------

for ((i=0; i<b; i++))
do
    blocks[$i]=${original[$i]}
done

echo ""
echo "--- First Fit ---"
echo "Process  Size  Block"

for ((i=0; i<p; i++))
do
    allocated=0

    for ((j=0; j<b; j++))
    do
        if [ ${blocks[$j]} -ge ${process[$i]} ]
        then
            echo "P$((i+1))       ${process[$i]}    B$((j+1))"
            blocks[$j]=$((blocks[$j] - process[$i]))
            allocated=1
            break
        fi
    done

    if [ $allocated -eq 0 ]
    then
        echo "P$((i+1))       ${process[$i]}    Not Allocated"
    fi
done

# ---------------- BEST FIT ----------------

for ((i=0; i<b; i++))
do
    blocks[$i]=${original[$i]}
done

echo ""
echo "--- Best Fit ---"
echo "Process  Size  Block"

for ((i=0; i<p; i++))
do
    best=-1

    for ((j=0; j<b; j++))
    do
        if [ ${blocks[$j]} -ge ${process[$i]} ]
        then
            if [ $best -eq -1 ] || [ ${blocks[$j]} -lt ${blocks[$best]} ]
            then
                best=$j
            fi
        fi
    done

    if [ $best -ne -1 ]
    then
        echo "P$((i+1))       ${process[$i]}    B$((best+1))"
        blocks[$best]=$((blocks[$best] - process[$i]))
    else
        echo "P$((i+1))       ${process[$i]}    Not Allocated"
    fi
done

# ---------------- WORST FIT ----------------

for ((i=0; i<b; i++))
do
    blocks[$i]=${original[$i]}
done

echo ""
echo "--- Worst Fit ---"
echo "Process  Size  Block"

for ((i=0; i<p; i++))
do
    worst=-1

    for ((j=0; j<b; j++))
    do
        if [ ${blocks[$j]} -ge ${process[$i]} ]
        then
            if [ $worst -eq -1 ] || [ ${blocks[$j]} -gt ${blocks[$worst]} ]
            then
                worst=$j
            fi
        fi
    done

    if [ $worst -ne -1 ]
    then
        echo "P$((i+1))       ${process[$i]}    B$((worst+1))"
        blocks[$worst]=$((blocks[$worst] - process[$i]))
    else
        echo "P$((i+1))       ${process[$i]}    Not Allocated"
    fi
done
```
