```python
%%writefile deadlock.c
#include <stdio.h>

int main() {
    int n, m;
    int allocation[10][10], request[10][10];
    int available[10], work[10];
    int finish[10], deadlocked[10];
    int count = 0;

    printf("Deadlock Detection Algorithm\n");
    printf("----------------------------\n");

    printf("Enter number of processes: ");
    scanf("%d", &n);

    printf("Enter number of resources: ");
    scanf("%d", &m);

    printf("\nEnter Allocation Matrix:\n");
    for (int i = 0; i < n; i++) {
        printf("P%d: ", i);
        for (int j = 0; j < m; j++) {
            scanf("%d", &allocation[i][j]);
        }
    }

    printf("\nEnter Request Matrix:\n");
    for (int i = 0; i < n; i++) {
        printf("P%d: ", i);
        for (int j = 0; j < m; j++) {
            scanf("%d", &request[i][j]);
        }
    }

    printf("\nEnter Available Resources:\n");
    for (int j = 0; j < m; j++) {
        scanf("%d", &available[j]);
        work[j] = available[j];
    }

    // Initialize finish array
    for (int i = 0; i < n; i++) {
        finish[i] = 0;
    }

    // Deadlock Detection Algorithm
    while (1) {
        int found = 0;

        for (int i = 0; i < n; i++) {
            if (finish[i] == 0) {
                int canFinish = 1;

                // Check Request[i] <= Work
                for (int j = 0; j < m; j++) {
                    if (request[i][j] > work[j]) {
                        canFinish = 0;
                        break;
                    }
                }

                if (canFinish) {
                    // Process can finish and release its resources
                    for (int j = 0; j < m; j++) {
                        work[j] += allocation[i][j];
                    }

                    finish[i] = 1;
                    found = 1;
                }
            }
        }

        if (found == 0) {
            break;
        }
    }

    // Find deadlocked processes
    for (int i = 0; i < n; i++) {
        if (finish[i] == 0) {
            deadlocked[count++] = i;
        }
    }

    if (count == 0) {
        printf("\nNo deadlock detected.\n");
        printf("All processes can complete successfully.\n");
    } else {
        printf("\nDEADLOCK DETECTED!\n");
        printf("Deadlocked Processes: ");

        for (int i = 0; i < count; i++) {
            printf("P%d", deadlocked[i]);

            if (i != count - 1) {
                printf(", ");
            }
        }

        printf("\n");
    }

    return 0;
}
```

```python
%%writefile deadlock.sh
#!/bin/bash

echo "Deadlock Detection Algorithm"
echo "----------------------------"

read -p "Enter number of processes: " n
read -p "Enter number of resources: " m

declare -A allocation
declare -A request
declare -a available
declare -a work
declare -a finish

echo ""
echo "Enter Allocation Matrix:"

for ((i=0; i<n; i++))
do
    echo -n "P$i: "

    for ((j=0; j<m; j++))
    do
        read allocation[$i,$j]
    done
done

echo ""
echo "Enter Request Matrix:"

for ((i=0; i<n; i++))
do
    echo -n "P$i: "

    for ((j=0; j<m; j++))
    do
        read request[$i,$j]
    done
done

echo ""
echo "Enter Available Resources:"

for ((j=0; j<m; j++))
do
    read available[$j]
    work[$j]=${available[$j]}
done

# Initialize finish array
for ((i=0; i<n; i++))
do
    finish[$i]=0
done

# Deadlock Detection
while true
do
    found=0

    for ((i=0; i<n; i++))
    do
        if [ ${finish[$i]} -eq 0 ]
        then
            canFinish=1

            for ((j=0; j<m; j++))
            do
                if [ ${request[$i,$j]} -gt ${work[$j]} ]
                then
                    canFinish=0
                    break
                fi
            done

            if [ $canFinish -eq 1 ]
            then
                # Release allocated resources
                for ((j=0; j<m; j++))
                do
                    work[$j]=$((work[$j] + allocation[$i,$j]))
                done

                finish[$i]=1
                found=1
            fi
        fi
    done

    if [ $found -eq 0 ]
    then
        break
    fi
done

# Check for deadlocked processes
deadlock=0

for ((i=0; i<n; i++))
do
    if [ ${finish[$i]} -eq 0 ]
    then
        if [ $deadlock -eq 0 ]
        then
            echo ""
            echo "DEADLOCK DETECTED!"
            echo -n "Deadlocked Processes: "
        fi

        echo -n "P$i "
        deadlock=1
    fi
done

echo ""

if [ $deadlock -eq 0 ]
then
    echo ""
    echo "No deadlock detected."
    echo "All processes can complete successfully."
fi
```
