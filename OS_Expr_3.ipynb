```python
%%writefile systemcalls.c
#include <stdio.h>
#include <stdlib.h>
#include <unistd.h>
#include <sys/wait.h>
#include <fcntl.h>

int main() {
    int fd;

    printf("Parent Process Started\n");
    printf("Parent PID: %d\n", getpid());

    // Open a file
    fd = open("sample.txt", O_CREAT | O_WRONLY, 0644);

    if (fd == -1) {
        printf("Error opening file\n");
        exit(1);
    }

    printf("File opened successfully\n");

    // Create child process
    pid_t pid = fork();

    if (pid < 0) {
        printf("Fork failed\n");
        close(fd);
        exit(1);
    }

    if (pid == 0) {
        // Child process
        printf("\nChild Process\n");
        printf("Child PID: %d\n", getpid());
        printf("Parent PID: %d\n", getppid());

        // Close file descriptor
        close(fd);
        printf("Child closed the file\n");

        // Exit child process
        exit(0);
    }
    else {
        // Parent process
        printf("\nParent is waiting for child...\n");

        // Wait for child
        wait(NULL);

        printf("Child process completed\n");

        // Close file descriptor
        close(fd);
        printf("Parent closed the file\n");

        printf("Parent process completed\n");
    }

    return 0;
}
```

```python
!gcc systemcalls.c -o systemcalls
!./systemcalls
```

```python
%%writefile systemcalls.sh
#!/bin/bash

echo "System Calls Demonstration"
echo "--------------------------"

echo "Shell Process ID: $$"

# Create a child/background process
(
    echo "Child Process Started"
    echo "Child Process ID: $BASHPID"
    sleep 2
    echo "Child Process Completed"
    exit 0
) &

child_pid=$!

echo "Parent Process ID: $$"
echo "Child Process ID: $child_pid"

echo "Parent is waiting for child..."

# Wait for child process
wait $child_pid

echo "Child process has completed"

# Create and close a file
echo "Creating file..."
exec 3>sample.txt

echo "Hello from Shell Script" >&3

echo "Closing file..."
exec 3>&-

echo "File closed"

echo "Shell script completed"
```
