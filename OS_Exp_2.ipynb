```python
%%writefile unix.c
#include <stdio.h>
#include <stdlib.h>

int main() {
    printf("UNIX Commands Demonstration\n");
    printf("---------------------------\n");

    printf("\nCurrent Directory:\n");
    system("pwd");

    printf("\nFiles and Directories:\n");
    system("ls");

    printf("\nDetailed File List:\n");
    system("ls -l");

    printf("\nCreating directory...\n");
    system("mkdir -p unix_demo");

    printf("\nCreating file...\n");
    system("touch unix_demo/example.txt");

    printf("\nWriting into file...\n");
    system("echo 'Hello from C program' > unix_demo/example.txt");

    printf("\nDisplaying file contents:\n");
    system("cat unix_demo/example.txt");

    return 0;
}
```

```python
%%writefile demo.sh
#!/bin/bash

echo "Shell Scripting Demonstration"
echo "-----------------------------"

echo "Enter your name:"
read name

echo "Hello, $name"

echo "Enter two numbers:"
read a
read b

sum=$((a + b))

echo "Sum = $sum"

if [ $sum -gt 10 ]
then
    echo "Sum is greater than 10"
else
    echo "Sum is 10 or less"
fi

echo "Numbers from 1 to 5:"

for i in 1 2 3 4 5
do
    echo $i
done
```
