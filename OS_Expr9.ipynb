```python
%%writefile threading.c
#include <stdio.h>
#include <stdlib.h>
#include <pthread.h>
#include <unistd.h>

// Function executed by Thread 1
void *thread1_function(void *arg) {
    printf("Thread 1 started\n");

    for (int i = 1; i <= 5; i++) {
        printf("Thread 1: %d\n", i);
        sleep(1);
    }

    printf("Thread 1 completed\n");

    return NULL;
}

// Function executed by Thread 2
void *thread2_function(void *arg) {
    printf("Thread 2 started\n");

    for (int i = 1; i <= 5; i++) {
        printf("Thread 2: %d\n", i);
        sleep(1);
    }

    printf("Thread 2 completed\n");

    return NULL;
}

int main() {
    pthread_t thread1, thread2;

    printf("Main thread started\n");

    // Create threads
    pthread_create(&thread1, NULL, thread1_function, NULL);
    pthread_create(&thread2, NULL, thread2_function, NULL);

    // Wait for threads to complete
    pthread_join(thread1, NULL);
    pthread_join(thread2, NULL);

    printf("All threads completed\n");

    return 0;
}
```

```python
!gcc threading.c -o threading -pthread
!./threading
```

```python
%%writefile threading.sh
#!/bin/bash

echo "Threading / Parallel Execution"
echo "-----------------------------"

# Function 1
thread1() {
    echo "Thread 1 started"

    for i in 1 2 3 4 5
    do
        echo "Thread 1: $i"
        sleep 1
    done

    echo "Thread 1 completed"
}

# Function 2
thread2() {
    echo "Thread 2 started"

    for i in 1 2 3 4 5
    do
        echo "Thread 2: $i"
        sleep 1
    done

    echo "Thread 2 completed"
}

# Run functions in background
thread1 &
PID1=$!

thread2 &
PID2=$!

echo "Both threads are running..."

# Wait for both background processes
wait $PID1
wait $PID2

echo "All threads completed"
```

```python
!chmod +x threading.sh
!./threading.sh
```
