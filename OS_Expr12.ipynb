```python
%%writefile page_replacement.c
#include <stdio.h>

int main() {
    int pages[50], frames[10];
    int n, f;
    int pageFaults = 0;
    int index = 0;

    printf("FIFO Page Replacement Algorithm\n");
    printf("--------------------------------\n");

    printf("Enter number of pages: ");
    scanf("%d", &n);

    printf("Enter reference string:\n");
    for (int i = 0; i < n; i++) {
        scanf("%d", &pages[i]);
    }

    printf("Enter number of frames: ");
    scanf("%d", &f);

    // Initialize frames
    for (int i = 0; i < f; i++) {
        frames[i] = -1;
    }

    printf("\nPage\tFrames\t\tStatus\n");

    for (int i = 0; i < n; i++) {
        int found = 0;

        // Check if page is already present
        for (int j = 0; j < f; j++) {
            if (frames[j] == pages[i]) {
                found = 1;
                break;
            }
        }

        if (found == 0) {
            // Page fault - replace the oldest page
            frames[index] = pages[i];
            index = (index + 1) % f;
            pageFaults++;

            printf("%d\t", pages[i]);

            for (int j = 0; j < f; j++) {
                printf("%d ", frames[j]);
            }

            printf("\tPage Fault\n");
        } else {
            printf("%d\t", pages[i]);

            for (int j = 0; j < f; j++) {
                printf("%d ", frames[j]);
            }

            printf("\tHit\n");
        }
    }

    printf("\nTotal Page Faults = %d\n", pageFaults);
    printf("Total Page Hits = %d\n", n - pageFaults);

    return 0;
}
```

```python
%%writefile page_replacement.sh
#!/bin/bash

echo "FIFO Page Replacement Algorithm"
echo "--------------------------------"

read -p "Enter number of pages: " n

echo "Enter reference string:"
read -a pages

read -p "Enter number of frames: " f

declare -a frames
index=0
pageFaults=0

# Initialize frames
for ((i=0; i<f; i++))
do
    frames[$i]=-1
done

echo ""
echo -e "Page\tFrames\t\tStatus"

for ((i=0; i<n; i++))
do
    page=${pages[$i]}
    found=0

    # Check whether page is already in frame
    for ((j=0; j<f; j++))
    do
        if [ ${frames[$j]} -eq $page ]
        then
            found=1
            break
        fi
    done

    if [ $found -eq 0 ]
    then
        # Page fault
        frames[$index]=$page
        index=$(((index + 1) % f))
        pageFaults=$((pageFaults + 1))

        echo -ne "$page\t"

        for ((j=0; j<f; j++))
        do
            echo -ne "${frames[$j]} "
        done

        echo -e "\tPage Fault"
    else
        echo -ne "$page\t"

        for ((j=0; j<f; j++))
        do
            echo -ne "${frames[$j]} "
        done

        echo -e "\tHit"
    fi
done

pageHits=$((n - pageFaults))

echo ""
echo "Total Page Faults = $pageFaults"
echo "Total Page Hits = $pageHits"
```
